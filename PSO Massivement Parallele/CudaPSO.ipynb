```python
import subprocess
import pandas as pd
import matplotlib.pyplot as plt
import re
from IPython.display import display

# Noms des 5 fonctions de benchmark définies dans le code CUDA
fonctions = [
    "Levy 3-dimensional",
    "Shifted Rastrigin",
    "Shifted Rosenbrock",
    "Shifted Griewank",
    "Shifted Sphere"
]

resultats = []

for i in range(5):
    print(f"Évaluation de la fonction {i} : {fonctions[i]}...")

    with open('kernel.h', 'r') as file:
        lignes = file.readlines()

    with open('kernel.h', 'w') as file:
        for ligne in lignes:
            if ligne.startswith("const int SELECTED_OBJ_FUNC"):
                file.write(f"const int SELECTED_OBJ_FUNC = {i};\n")
            else:
                file.write(ligne)

    subprocess.run(["nvcc", "-o", "programpso", "main.cpp", "kernel.cpp", "kernel.cu"], check=True)

    process = subprocess.run(["./programpso"], capture_output=True, text=True)
    output = process.stdout
    match = re.search(r'GPU\s+([0-9.]+)\s+([0-9.\-eE]+)', output)

    if match:
        temps = float(match.group(1))
        minimum = float(match.group(2))
        resultats.append({
            "Fonction (ID)": f"{fonctions[i]} ({i})",
            "Temps d'exécution (s)": temps,
            "Minimum trouvé": minimum
        })
    else:
        print(f"Erreur de lecture pour la fonction {i} :\n{output}")
df = pd.DataFrame(resultats)
print("\n=== RÉSULTATS DES BENCHMARKS ===")
display(df)
plt.figure(figsize=(10, 6))
plt.bar(df['Fonction (ID)'], df['Minimum trouvé'], color='coral', edgecolor='black')
plt.yscale('symlog')
plt.title('Comparaison des minimums trouvés par fonction ')
plt.ylabel('Minimum (Échelle SymLog)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Création du graphique pour les temps d'exécution
plt.figure(figsize=(10, 6))
plt.bar(df['Fonction (ID)'], df['Temps d\'exécution (s)'], color='skyblue', edgecolor='black')

plt.title('Comparaison des temps d\'exécution par fonction')
plt.ylabel('Temps (secondes)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()

plt.show()
```
